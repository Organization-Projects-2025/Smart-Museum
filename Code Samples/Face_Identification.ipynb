```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
img_path = "/content/abba.png"
module_path = "/content/haarcascade_frontalface_default.xml"
HaarCascade = cv2.CascadeClassifier(module_path)
```

```python
img_open = cv2.imread(img_path)
cv2_imshow(img_open)
```

```python
img_gray = cv2.cvtColor(img_open, cv2.COLOR_BGR2GRAY)
faces = HaarCascade.detectMultiScale(img_gray, minSize=(30, 30))
print(len(faces))
```

```python
for x, y, w, h in faces:
    cv2.rectangle(img_open, (x, y), (x + w, y + h), (0, 255, 0), 2)
cv2_imshow(img_open)
```

```python
img_open = cv2.imread("/content/retinaface-results.jpeg")
# cv2_imshow(img_open)
img_gray = cv2.cvtColor(img_open, cv2.COLOR_BGR2GRAY)
faces = HaarCascade.detectMultiScale(img_gray, minSize=(30, 30))
for x, y, w, h in faces:
    cv2.rectangle(img_open, (x, y), (x + w, y + h), (0, 255, 0), 2)
cv2_imshow(img_open)
```
